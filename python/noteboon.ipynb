```python
import csv
import json
import re
import xml.etree.ElementTree as ET
from bs4 import BeautifulSoup

print("PARSING TEXT FILE")
print("-" * 40)

# 1️⃣ Parsing Text
with open("data.txt", "w") as f:
    f.write("Ali 85\nAhmed 45\nSara 120")

with open("data.txt", "r") as f:
    for line in f:
        name, marks = line.split()
        print("Name:", name, "Marks:", marks)

print("\nPARSING CSV FILE")
print("-" * 40)

# 2️⃣ Parsing CSV
with open("data.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["Name", "Marks"])
    writer.writerow(["Ali", 85])
    writer.writerow(["Ahmed", 45])
    writer.writerow(["Sara", 120])

with open("data.csv", "r") as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row["Name"], row["Marks"])

print("\nPARSING HTML")
print("-" * 40)

# 3️⃣ Parsing HTML
html = "<html><body><h1>Ali</h1><p>85</p></body></html>"
soup = BeautifulSoup(html, "html.parser")
print("Name:", soup.h1.text)
print("Marks:", soup.p.text)

print("\nPARSING XML")
print("-" * 40)

# 4️⃣ Parsing XML
xml_data = """<student>
<name>Ali</name>
<marks>85</marks>
</student>"""

root = ET.fromstring(xml_data)
print("Name:", root.find("name").text)
print("Marks:", root.find("marks").text)

print("\nPARSING JSON")
print("-" * 40)

# 5️⃣ Parsing JSON
json_data = '{"name": "Ali", "marks": 85}'
data = json.loads(json_data)
print("Name:", data["name"])
print("Marks:", data["marks"])

print("\nDETECTING DATA ANOMALIES")
print("-" * 40)

# 6️⃣ Detecting Anomalies
marks_list = [85, 45, 120, -5]

for mark in marks_list:
    if mark < 0 or mark > 100:
        print("Anomaly Detected:", mark)
    else:
        print("Valid Marks:", mark)
```
